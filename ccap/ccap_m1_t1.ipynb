```python
from google.colab import userdata
from google import genai

client = genai.Client(
    api_key=userdata.get("GOOGLE_API_KEY")
)

response = client.models.generate_content(
    model="gemini-3.8-flash",
    contents="Explain what RAG is in simple English."
)

print(response.text)
```
